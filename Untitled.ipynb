```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

# Create sample data
dates = [datetime(2024, 1, 1) + timedelta(days=x) for x in range(90)]
regions = ['North', 'South', 'East', 'West']
products = ['Product A', 'Product B', 'Product C', 'Product D']

data = []
for date in dates:
    for region in regions:
        for product in products:
            data.append({
                'Date': date,
                'Region': region,
                'Product': product,
                'Sales': np.random.randint(1000, 5000),
                'Units_Sold': np.random.randint(10, 100),
                'Customer_Count': np.random.randint(5, 50)
            })

df = pd.DataFrame(data)
df.to_excel('sample_sales_data.xlsx', index=False)
print("Sample data created: sample_sales_data.xlsx")
print(df.head())
```

```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

np.random.seed(42)

customer_ids = [f"CUST_{i:04d}" for i in range(1, 201)]
regions = ['North', 'South', 'East', 'West']
tiers = ['Bronze', 'Silver', 'Gold', 'Platinum']
channels = ['Online', 'Retail Store', 'Phone']

data = []
start_date = datetime(2024, 1, 1)

for cust in customer_ids:
    region = np.random.choice(regions, p=[0.3, 0.3, 0.2, 0.2])
    tier = np.random.choice(tiers, p=[0.4, 0.3, 0.2, 0.1])
    channel = np.random.choice(channels)
    
    # Number of orders varies, with Platinum customers ordering more
    if tier == 'Platinum':
        n_orders = np.random.randint(15, 30)
        order_value_base = np.random.randint(300, 600)
    elif tier == 'Gold':
        n_orders = np.random.randint(8, 15)
        order_value_base = np.random.randint(150, 300)
    else:
        n_orders = np.random.randint(1, 8)
        order_value_base = np.random.randint(30, 150)
    
    for _ in range(n_orders):
        days_offset = np.random.randint(0, 180)
        order_date = start_date + timedelta(days=days_offset)
        # South region orders trail off after day 100 (simulated churn pattern)
        if region == 'South' and days_offset > 100:
            if np.random.random() > 0.3:
                continue
        order_value = order_value_base + np.random.randint(-50, 100)
        data.append({
            'Customer_ID': cust,
            'Region': region,
            'Tier': tier,
            'Channel': channel,
            'Order_Date': order_date,
            'Order_Value': max(order_value, 10)
        })

df = pd.DataFrame(data)
df.to_excel('customer_data.xlsx', index=False)
print("Created customer_data.xlsx")
print(df.shape)
```
