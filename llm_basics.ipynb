```python
from dotenv import load_dotenv
import os
from openai import OpenAI

load_dotenv()
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
```

```python
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "You are a sarcastic data scientist who explains things simply."},
        {"role": "user", "content": "Explain what a transformer model is in 3 sentences."}
    ]
)
print(response.choices[0].message.content)
```

```python
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "You are a helpful data science assistant."},
        {"role": "user", "content": "What is a transformer model?"},
        {"role": "assistant", "content": "A transformer model is..."},
        {"role": "user", "content": "How is it different from an RNN?"}
    ]
)
print(response.choices[0].message.content)
```

```python
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "user", "content": "What is machine learning in one sentence?"}
    ]
)
print(response.choices[0].message.content)
print("\nTokens used:")
print("Prompt tokens:", response.usage.prompt_tokens)
print("Completion tokens:", response.usage.completion_tokens)
print("Total tokens:", response.usage.total_tokens)
```

```python
def ask_data_assistant(question, conversation_history = []):
    conversation_history.append({
    "role": "user",
    "content": question
    })
    
    response = client.chat.completions.create(
        model = "gpt-4o-mini", 
        messages = [{"role": "system", "content": "You are a helpful data science assistant. Keep answers concise."}] + conversation_history
    )
    answer = response.choices[0].message.content
    conversation_history.append({
        "role": "assistant", 
        "content": answer
    })
    
    return answer, conversation_history

#Testing it
history = []

answer, history = ask_data_assistant("What is a vector database?",history)
print("Answer 1: ",answer)

answer, history = ask_data_assistant("How is it different from a regular database?",history)
print("\nAnswer 2: ",answer)

answer, history = ask_data_assistant("Give me an example of when I'd use it",history)
print("\nAnswer 3: ",answer)
```
